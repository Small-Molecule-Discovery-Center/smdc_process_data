# Extract m/z and intensity data from Xevo/MassLynx .rpt files
This notebook will extract m/z and intensity data from .rpt files into text files suitable for use with other graphing software such as [ChartographMS](http://chartograph.com/ms/index.html).
- Each sample is extracted into its own text file in the folder `samples` and named with the `fname` corresponding to the sample of interest.
- An index file is created called "extracted_samples.csv" with information about the plate, well, sample name, and max values for intensity and mass of the max intensity. This is also located in the `samples` folder.
- The whole `samples` folder is zipped for easy download as `samples.zip`.
